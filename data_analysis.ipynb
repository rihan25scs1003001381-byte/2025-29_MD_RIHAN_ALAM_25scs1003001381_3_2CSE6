# Initial Data Exploration — Sample Sales Dataset

This notebook is used for quick, initial exploration of `data/sample_sales.csv` before the logic is turned into reusable functions in `data_analyzer.py` and `visualizer.py`.

Workflow followed:

`Data Collection -> Data Cleaning -> Exploratory Data Analysis -> Data Visualization -> Question Analysis -> Insights`

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
%matplotlib inline
```

## 1. Data Collection — load the dataset

```python
df = pd.read_csv("../data/sample_sales.csv")
print("Shape:", df.shape)
df.head()
```

```python
df.info()
```

## 2. Data Cleaning — check for issues

```python
# Missing values per column
df.isnull().sum()
```

```python
# Duplicate rows
print("Duplicate rows:", df.duplicated().sum())
df[df.duplicated(keep=False)]
```

```python
# Numeric vs categorical columns
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
categorical_cols = df.select_dtypes(exclude=np.number).columns.tolist()
print("Numeric columns:", numeric_cols)
print("Categorical columns:", categorical_cols)
```

## 3. Exploratory Data Analysis

```python
df.describe()
```

```python
df["Category"].value_counts()
```

```python
revenue_by_product = df.groupby("Product")["Revenue"].sum().sort_values(ascending=False)
revenue_by_product
```

## 4. Data Visualization

```python
fig, ax = plt.subplots(figsize=(8, 5))
sns.barplot(x=revenue_by_product.values, y=revenue_by_product.index, ax=ax, palette="viridis")
ax.set_xlabel("Total Revenue")
ax.set_title("Revenue by Product")
plt.tight_layout()
plt.show()
```

```python
corr = df.select_dtypes(include=np.number).corr()
fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", ax=ax)
ax.set_title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

## 5. Question Analysis

Example: *"Which product generated the highest revenue?"*

```python
top_product = revenue_by_product.idxmax()
top_revenue = revenue_by_product.max()
print(f"'{top_product}' generated the highest revenue: {top_revenue}")
```
